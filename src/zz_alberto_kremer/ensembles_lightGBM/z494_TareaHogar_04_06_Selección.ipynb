## 3. Analisis de los modelos

#### Ahora que tengo un "dataset" con los resultados de los experimentos en los 12 modelos, puedo proceder a un análisis estadístico.

#### Calcularé: AUC medio, mediana, desvío estandar, IQC y AUC mínimo y mínimo para cada uno de los 12 modelos a evaluar.

```r
tb_resumen <- tb_resultados[,.(
    AUC_medio   = mean(AUC),
    AUC_mediano = median(AUC),
    SD_AUC      = sd(AUC),
    IQR_AUC     = IQR(AUC),
    AUC_min     = min(AUC),
    AUC_max     = max(AUC)
  ), by = modelo]

setorder(tb_resumen, -AUC_medio)

tb_resumen
```

#### En un primer análisis, ordenando por AUC medio de mayor a menor, surge que los 5 mejores modelos son: M02, M03, M04, M05 y M06.

## 4. Comparación entre los modelos

### 4.1 Aplico Wilcoxon para comparar los 12 modelos, todos contra todos.

```r
# Comparación de los 12 modelos mediante Wilcoxon pareado

modelos <- sort(unique(tb_resultados$modelo))

tb_wilcoxon <- rbindlist(
  lapply(
    combn(modelos, 2, simplify = FALSE),
    function(x) {

      m1 <- x[1]
      m2 <- x[2]

      auc1 <- tb_resultados[modelo == m1][order(par), AUC]
      auc2 <- tb_resultados[modelo == m2][order(par), AUC]

      prueba <- wilcox.test(
        auc1, auc2,
        paired = TRUE,
        exact = TRUE
      )

      media1 <- mean(auc1)
      media2 <- mean(auc2)

      ganador <- if (prueba$p.value < 0.05) {
        if (media1 > media2) m1 else m2
      } else {
        NA_character_
      }

      data.table(
        modelo1 = m1,
        modelo2 = m2,
        p_value = prueba$p.value,
        ganador = ganador
      )
    }
  )
)

# Cantidad de comparaciones Wilcoxon ganadas significativamente

tb_seleccion <- data.table(modelo = modelos)

tb_victorias <- tb_wilcoxon[
  !is.na(ganador),
  .(victorias_wilcoxon = .N),
  by = .(modelo = ganador)
]

tb_seleccion <- merge(
  tb_seleccion,
  tb_victorias,
  by = "modelo",
  all.x = TRUE
)

tb_seleccion[
  is.na(victorias_wilcoxon),
  victorias_wilcoxon := 0L
]

# AUC medio sólo como criterio de desempate

tb_medias <- tb_resultados[
  ,
  .(AUC_medio = mean(AUC)),
  by = modelo
]

tb_seleccion <- merge(
  tb_seleccion,
  tb_medias,
  by = "modelo"
)

setorder(
  tb_seleccion,
  -victorias_wilcoxon,
  -AUC_medio
)

tb_seleccion
```

### En función de los resultados al comparar por Wilcoxon, se mantiene la situación que nos permitía seleccionar M02, M03, M04, M05, M06.

### Por tal motivo, serán los modelos que enviaré a Kaggle.
