```r
# ============================================================
# EJECUCIÓN SECUENCIAL DE LOS 7 NOTEBOOKS - ESCENARIO 1
# ============================================================

notebooks <- c(
  "z620_Esc1_9111_700001.ipynb",
  "z620_Esc1_9112_690506.ipynb",
  "z620_Esc1_9113_804528.ipynb",
  "z620_Esc1_9114_613037.ipynb",
  "z620_Esc1_9115_648983.ipynb",
  "z620_Esc1_9116_365305.ipynb",
  "z620_Esc1_9117_163687.ipynb"
)

for (notebook in notebooks) {

  cat("\n")
  cat("============================================================\n")
  cat("INICIANDO:", notebook, "\n")
  cat("============================================================\n")

  salida <- system2(
    "jupyter",
    args = c(
      "nbconvert",
      "--to", "notebook",
      "--execute",
      "--ExecutePreprocessor.timeout=-1",
      "--inplace",
      shQuote(notebook)
    ),
    stdout = TRUE,
    stderr = TRUE
  )

  estado <- attr(salida, "status")

  if (is.null(estado)) {
    estado <- 0
  }

  cat(paste(salida, collapse = "\n"))
  cat("\n")

  if (estado != 0) {

    stop(
      paste(
        "ERROR al ejecutar",
        notebook,
        "- se detiene la secuencia."
      )
    )

  }

  cat("\n")
  cat("FINALIZADO:", notebook, "\n")
  cat("============================================================\n")
}

cat("\n")
cat("============================================================\n")
cat("TODOS LOS NOTEBOOKS FUERON EJECUTADOS CORRECTAMENTE.\n")
cat("============================================================\n")
```
